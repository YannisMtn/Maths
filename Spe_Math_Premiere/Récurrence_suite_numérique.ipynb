## Suites arithémtiques

### L'objectif est de se familiariser avec les outils mathématiques et modéliser des objets mathématiques.

>Dans un premier temps il faut importer les librairies nécessaires au fonctionnement. SymPy est le meilleur choix pour la phase de généralisation et de démonstration.

Le théorème à démontrer est : 

$ \text{Une suite arithmétique est entièrement et uniquement déterminée par la donnée de sa raison r et de son premier terme } u_0. \text{On a alors pour tout }n \geq 0  : $
$$u_n=u_0+nr$$

```python
import sympy
```

Maintenant il faut poser les variables et la suite.

```python
from sympy import symbols, Function
```

```python
n= symbols('n', integer= True, nonnegative=True)
r = symbols('r', real=True)
u = Function('u')

suite = u(0)+ n * r
```

La suite est posée, il faut maintenant faire la démonstration.

# Initialisation : 

Je vérifie si la propriété est vraie pour le premier terme, soit n=0

```python
from sympy import Eq
# j'importe la fonction qui me permet de vérifier des égalités
```

```python
Eq((u(0)), suite.subs(n, 0))
#j'effectue l'égalité entre u(0) la valeur que je suppose obtenir (cela correspond à u_n avec n=0, soit u(0)) et celle obtenue dans la fonction définie lorsque n=0.
```

L'initialisation est vérifiée. je peux passer à la définition de l'hypothèse de récurrence.

# Hypothèse de récurrence :

Supposons la propriété $P(n)$ vérifiée $ \forall n \geq 0 $. Montrons que cette hypothèse entraîne la validité de $P(n+1)$ soit :

```python
récurrence = u(0) + (n+1) * r
```

## Hérédité
Je sais que pour qu'une suite soit considérée comme arithmétique elle doit respecter la condition suivante : 
$$u_{n+1} = u_n + r$$

alors je dois obtenir $u_{n+1} - u_{n} = r$

Je vais donc utiliser l'hypothèse de récurrence $u_{n+1} = u_0 + (n+1) * r$ avec la différence énoncée précédemment.

```python
from sympy import expand

display(expand(récurrence))
display(expand(suite))

résultat = (expand(récurrence) - expand(suite))
```

```python
Eq(r , résultat)
```

## Conclusion

La propriété est vraie au terme initial et héréditaire, donc par le principe de récurrence $P(n)$ est vraie $\forall n \in \mathbb{N}, n \geq 0$

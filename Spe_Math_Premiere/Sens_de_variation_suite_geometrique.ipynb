## Modélisation des variations d'une suite géométrique de raison q non nulle et de premier terme non nul.

> Premièrement il faut importer les librairies nécessaires à la réalisation de la suite ainsi qu'à sa modélisation.

```python
import sympy # pour les calculs symboliques
import matplotlib as plt # pour la visualisation
import numpy as np # pour les calculs numériques
```

- Je vais maintenant poser via sympy la suite u(n) géométrique.
- Utiliser numpy pour le passage aux valeurs numériques en enregistrant les valeurs de la suite dans une liste ou un vecteur.
- Enfin préparer le graphique et afficher les valeurs obtenues en fonction de chaque paramètre pour voir les différents comportements des suites.

```python
from sympy import symbols, Function
```

```python
n = symbols('n', integer=True)
u0 = symbols('u0', real=True)
q = symbols('q', real=True)

suite = u0 * q**n
```

```python
from sympy import lambdify # pour transformer une expression symbolique en fonction numérique
```

```python
f_numerique = lambdify((u0, q, n), suite, 'numpy')

def valeur_numeriques_suite(u0_choisi, q_choisi, n_max): 
    list_n = np.arange(0, n_max + 1)
    valeur_suite = f_numerique(u0_choisi, q_choisi, list_n)    

    return list_n, valeur_suite 

# demande des valeurs à l'utilisateur
u0_user = float(input("Entrez la valeur de u0 : "))
q_user = float(input("Entrez la valeur de q : "))
n_user = int(input("Entrez la valeur de n : "))

vecteur_n, vecteur_u_n = valeur_numeriques_suite(u0_user, q_user, n_user)

# affichage des résultats
print("\nVecteur des n :", vecteur_n)
print("Vecteur des u_n :", vecteur_u_n)
```

Maintenant il faut faire le graphique.

```python
import matplotlib.pyplot as plt
plt.figure(figsize=(10, 6))

# On trace les points (vecteur_n en X, vecteur_u_n en Y)
plt.stem(vecteur_n,vecteur_u_n, linefmt='b-', markerfmt='bo', basefmt='r-')

# Habillage du graphique
plt.title(f"Comportement de la suite géométrique (u0={u0_user}, q={q_user})")
plt.xlabel("Indice n")
plt.ylabel("Valeurs de u_n")
plt.grid(True)

# Affichage
plt.show()
```

Je peux maintenant bien observer les variations des suites en fonction de leur premier terme et leur raison.
